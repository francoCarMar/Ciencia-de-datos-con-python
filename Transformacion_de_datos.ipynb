```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

filename = 'data/pima-indians-diabetes.csv'
names = ['preg', 'plas', 'pres', 'skin', 'test', 'mass', 'pedi', 'age', 'class' ]
data = pd.read_csv(filename, names=names)
array = data.values
X = array[:,0:8]
Y = array[:,8]
Y
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 1))
rescaledX = scaler.fit_transform(X)
np.set_printoptions(precision=3)
print(names)
print(rescaledX[0:5,:])
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler().fit(X)
rescaledX = scaler.transform(X)
np.set_printoptions(precision=3)
print(names)
print(rescaledX[0:5,:])
```

```python
# Normalizar data
from sklearn.preprocessing import Normalizer
scaler = Normalizer().fit(X)
normalized_X = scaler.transform(X)
print(names)
print(normalized_X[0:5, :])
```

```python
# binarizacion
from sklearn.preprocessing import Binarizer
binarizer = Binarizer(threshold=0.0).fit(X)
binarized_X = binarizer.transform(X)
print(names)
print(binarized_X[0:5, :])
```

```python
# Box - Cox
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20, 20))

plt.subplot(421)
sns.histplot(data['pedi'], kde=True)
plt.title('Pedigree')
plt.grid(True)

plt.subplot(422)
sns.histplot(data['age'], kde=True)
plt.title('Age')
plt.grid(True)

plt.subplot(423)
sns.histplot(data['mass'], kde=True)
plt.title('Mass')
plt.grid(True)

plt.subplot(424)
sns.histplot(data['class'], kde=False)  # class parece categórica
plt.title('Class')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import PowerTransformer
# extraer caracteristicas con sesgo
features = data[['pedi', 'age']]
pt = PowerTransformer(method='box-cox', standardize=True)
skl_boxcox = pt.fit(features)
calc_lambdas = skl_boxcox.lambdas_
skl_boxcox = skl_boxcox.transform(features)
df_features = pd.DataFrame(data = skl_boxcox, columns = ['pedi', 'age'])

# Pasar las caracteristicas transformadas al dataset original
data.drop(['pedi'], axis = 1, inplace = True)
data.drop(['age'], axis = 1, inplace = True)
df_data = pd.concat([data, df_features], axis = 1)
cols = df_data.columns.tolist()
cols = cols[-1:] + cols[:-1]
cols = cols[-1:] + cols[:-1]
df_data = df_data[cols]
df_data
```

```python
plt.figure(figsize=(20, 20))

plt.subplot(421)
sns.histplot(df_data['pedi'], kde=True)
plt.title('Pedigree')
plt.grid(True)

plt.subplot(422)
sns.histplot(df_data['age'], kde=True)
plt.title('Age')
plt.grid(True)

plt.subplot(423)
sns.histplot(df_data['mass'], kde=True)
plt.title('Mass')
plt.grid(True)

plt.subplot(424)
sns.histplot(df_data['class'], kde=False)  # class parece categórica
plt.title('Class')
plt.grid(True)

plt.tight_layout()
plt.show()
```
